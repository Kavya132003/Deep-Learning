**Major Differences between Machine Learning and Deep learning**

**Machine Learning**
* Includes mathematical operations.
* Direct implementation of Algebra, Probability, Statistics.
* Tabular numerical data
* It doesnt have capacity for internal conversion.
* It works best for small datasets.
* It needs a huge resources like hardware support(CPU, GPU, RAM).
* Types of Machine Learning:
  * Supervised Learning
  * Unsupervised Learning
  * Reinforcement Learning
* It does not have self learning capacity (Only Forward Propagation)
* Sklearn


**Deep Learning**
* Biological neurons (Human Brains)
* Derivatives and Integrals
* Tabular data, Audio, Video, Image, Text.
* It have capacity to convert internally, any data into numerical format.
* It works best with large datasets.
* Less resources comparatively.
* Types of Deep Learning:
  * Artificial Neural Network(ANN)
  * Convolutional Neural Network(CNN)
  * Recurrent Neural Networks(RNN)
* It have self-learning capacity (both backward and forward propagation)
* TensorFlow, PyTorch

**Deep Learning Architecture**
1) Single Layer Perceptron (Single-Layered Neural Network)
* It is the simplest type of Artificial neural network.
* It is a linear binary classifier used for supervised learning.
* It has one input layer(no hidden layers)  and one output layer.

**Architectural diagram**



* weighted sum (net input)


**Working Procedure**
* Initialize weights and bias with small random values.
* For each training input, it calculates the output y.
* If the predicted y is not equal to the actual y, then it updates the weights and bias using the perceptron update rule.
* Repeat until all training inputs are classified correctly or the maximum iterations are reached.

**Properties of single-layered perceptron**
* It is used only for binary classification.
* It can solve linearly separable problems.
* It converges in finite steps if the data is linearly separable.
* It cannot solve non-linearly separable problems.

**Applications**
* Pattern recognition
* Spam detection
* Simple Binary Classification task


2) Multilayer Perceptron (Multi-Layered Neural network)
* It is a type of feed-forward artificial neural network consisting of an input layer, one or more hidden layers, and an output layer.
* It uses non-linear activation functions in hidden layers allowing it to learn complex patterns and solve non-linear problems.

**Multi-Layered Perceptron Architecture**



* Inputs(xi) -> feature values from the data.
* Weights(wi) -> It is a parameter representing the strength of the connection to the input.
* Bias(b) -> It is an additional parameter that shifts the activation function, allowing the neuron to fire even if all inputs are zero.
* Weighted sum(Net input) -> The linear combination of inputs and weights+bias.
* Activation function -> A function applied to the weighted sum to determine the neuron's output. In the original perceptron, this was the simple step function that decides whether the neuron has to activate or not based on the activation value.
* Types of Layers
  * Input Layer -> Receives the raw input features
  * Hidden Layer -> One or more layers between the input and output. Neurons in these layers use non-linear activation functions which is key to allowing the multilayer perceptron to learn and model non-linear relationships in the data.
  * Output Layer -> Produces the final result; It gives probability output for classification and a continuous value for regression. Here, the activation function is chosen based on the task.

* Input -> Output => Forward Propagation
* Output -> 1st Hidden Layer => Backward Propagation
* One forward propagation + One backward propagation = 1 epoch/ 1 iteration

* When loss value is high then only the backward propagation will happen.


**Forward propagation**
* It is a process where the input data is passed through the neural network layer by layer to compute the final output prediction



**29th July 2026**

**Backward Propagation**
* It is the algorithm used to efficiently calculate the gradients (derivatives) of the loss function with respect to every weight and bias in the network.
* These gradients tell us the direction and magnitude by which the parameter should be adjusted to minimize the loss.

**Gradient descent curve**


* There are two types of problems that occur for GDC
  * Vanishing Gradient Problem -> The vanishing gradient problem is the training difficulty in deep neural networks where gradients shrink close to 0 in earlier layers, stopping those layers from learning. It happens because of the chain rule, saturating activation functions and deep network depth.


  * Exploding Gradient Problem -> The exploding gradient problem is a training issue in deep neural networks where gradients grow exponentially large during backpropagation due to repeated multiplication of values > 1 via the chain rule, leading to unstable weight updates, model divergence or numerical overflow into null values.

 
**Weight Initialization Techniques**
*Why it is important?*
* Weights decide the starting point of learning.
* The goal is to keep the variance of the activations and gradients stable across all layers.

* **Good intialization**
* Faster convergence
* Better Performance

* **Bad initialization**
* Vanishing/ Exploding gradients lead to slow convergence

1) Zero Initialization:
* All weights are initialized to zero. w=0
  * Pros:
    * Simple
  * Cons:
    * All neurons will learn the same features.
    * Symmetry is not broken.
    * Not used in practice.

2) Random Initialization (Uniform):
* Weights are initialized from a uniform distribution in a small range.
* w ~ U(-a,a)
* where a is small (eg., 0.01 or 0.1)
  * Pros:
    * It breaks the symmetry
    * Easy to implement
  * Cons:
    * Not ideal for deep neural networks
    * May cause vanishing/exploding gradients

3) XAVIER/GLOROT Initialization:
* It works well with activation functions like tanh and sigmoid


* Pros:
    * Keeps variance of activations and gradients stable.
    * Good for tanh and sigmoid activation functions
  * Cons:
    * Not ideal for ReLU activation

4) He Initialization (Kaiming Initialization):
* It is designed for ReLU and its variants like Leaky ReLU, PReLU


* Pros:
    * Works well with ReLU and prevents the vanishing gradient problem
    * Better convergence with deep networks

5) Le Cun Initialization:
* It is mostly used with SELU(Scaled Exponential Linear Unit) activation function


* Pros:
    * Good for SELU activation
    * Keeps variance close to 1
